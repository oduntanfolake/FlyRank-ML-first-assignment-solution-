## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

url = "https://raw.githubusercontent.com/oduntanfolake/FlyRank-ML-first-assignment-solution-/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

df['ctr'] = df['clicks_90d'] / df['impressions_90d'].replace(0, 1)

total_pages = len(df)
avg_ctr = df['ctr'].mean()
under_cap_pct = ((df['avg_position'] <= 10) & (df['ctr'] < 0.03)).mean() * 100

print(" Quick look at the data")
print(f"Total pages: {total_pages}")
print(f"Average CTR: {avg_ctr:.4f}")
print(f"Percentage of Top 10 pages with CTR < 3%: {under_cap_pct:.2f}%")
```
