## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
print ("The main success metric for the ranking is Precision@50.\nPrecision@50 asks: **Of the 50 pages ranked highest by the engagement opportunity score, what proportion are genuinely relevant engagement opportunities?\n\nFor example, if 37 of the top 50 ranked pages are genuinely relevant opportunities:\nPrecision@50 = 37 / 50 = 0.74, or 74%.\n\nThis metric focuses on the quality of the pages at the top of the ranking because the website manager has limited time and needs to prioritize which pages to investigate first.\n\nA full evaluation would require an observed or independently validated relevance outcome. The current engagement-opportunity proxy is based on observed signals, so I would not treat a Precision@50 value calculated from the same proxy as independent evidence that the model is correct.")
```

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
print("A simple fixed rule such as \"flag every page with CTR below a certain threshold\" would only consider one signal at a time.\n\nThis lane uses multiple engagement signals, including clicks, CTR, impressions, and changes between recent and previous periods. These signals can provide different information about a page's engagement performance.\n\nA collection of hand-written rules could become complicated and may still fail to capture how these signals interact. An ML-based scoring approach could learn patterns across multiple signals and use them to prioritize pages rather than relying on one manually chosen threshold.\n\nThe output is decision support, not an automatic content change. A website manager uses the ranked pages and improvement areas to decide which pages to investigate and what action to take.\n\nIf the model ranks an unsuitable page highly, the main cost is that the manager may spend time investigating a page that did not need attention.")
```
